# Task A: Matrix Generation

```python
import numpy as np

# Set the NumPy random seed to 42 for reproducibility.
np.random.seed(42)

# Generate Matrix A of shape (4, 3) containing random integers in the range [1, 10].
A = np.random.randint(1, 11, size=(4, 3))

# Generate Matrix B of shape (3, 5) containing random integers in the range [1, 10].
B = np.random.randint(1, 11, size=(3, 5))

# Display both matrices with clear labels along with their .shape attributes.
print("Matrix A:")
print(A)
print(f"Shape of A: {A.shape}\n")

print("Matrix B:")
print(B)
print(f"Shape of B: {B.shape}")
```

# Task B: Dimension Validation & Dot Product

```python
# Programmatically validate that the inner dimensions match using an assertion:
assert A.shape[1] == B.shape[0], "Inner dimensions do not match for matrix multiplication!"

# Compute the matrix product: C = A * B
C = np.dot(A, B)

# Display the resulting matrix C and verify its shape is (4, 5).
print("Matrix C (A dot B):")
print(C)
print(f"Shape of C: {C.shape}")
assert C.shape == (4, 5), "Shape of C is incorrect!"
```

# Task C: Reductions & Summary Statistics

```python
# Compute and display the total sum of all elements in matrix C:
total_sum = np.sum(C)
print(f"Total sum of all elements in C: {total_sum}\n")

# Compute and display the row-wise sums ( axis=1 ).
row_sums = np.sum(C, axis=1)
print(f"Row-wise sums of C: {row_sums}\n")

# Compute and display the column-wise sums ( axis=0 ).
col_sums = np.sum(C, axis=0)
print(f"Column-wise sums of C: {col_sums}")
```
